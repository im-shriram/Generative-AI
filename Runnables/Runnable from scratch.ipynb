```python
from typing import Any
from abc import ABC, abstractmethod

class Runnable(ABC):
    def __init__(self) -> None:
        pass

    @abstractmethod
    def invoke(self, input: Any):
        pass
```

```python
class ChatModel(Runnable):
    def __init__(self, model: str, max_tokens: int, api_key: str) -> None:
        self.model = model

    def invoke(self, input: Any) -> dict[str, list]:
        response = {
            "content": "I'm an AI Assistant and I'm here to help you with anything.",
            "metadata": []
        }
        return response

class PromptTemplate(Runnable):
    def __init__(self, template: str, input_variables: list, validate_input: bool) -> None:
        self.template = template

    def invoke(self, input: Any):
        prompt: str = f"{self.template} {input}"
        return prompt

class StrOutputParser(Runnable):
    def __init__(self) -> None:
        pass

    def invoke(self, input: Any) -> str:
        return input['content']
```

```python
class Chain(Runnable):
    def __init__(self, runnables: list[Runnable]) -> None:
        self.runnables: list[Runnable] = runnables
        self.response = None
    
    def invoke(self, input: Any):
        self.response = input
        for runnable in self.runnables:
            self.response = runnable.invoke(input=self.response)

        return self.response
```

```python
template = PromptTemplate(template="Who are you?", input_variables=[], validate_input=False)
model = ChatModel(model="gemini-3-flash-preview", max_tokens=10, api_key="")
parser = StrOutputParser()

chain = Chain(runnables=[template, model, parser])
chain.invoke(input="Name")
```
